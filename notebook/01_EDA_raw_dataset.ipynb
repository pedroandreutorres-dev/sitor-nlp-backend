# Carga y Exploración Inicial de Datasets

```python
import os
import pandas as pd

# Creo un df_x para cada csv que se encuentra en data_files
df_01 = pd.read_csv("../data/raw/aa_dataset-tickets-multi-lang-5-2-50-version.csv")
df_02 = pd.read_csv("../data/raw/dataset-tickets-multi-lang-4-20k.csv")
df_03 = pd.read_csv("../data/raw/dataset-tickets-multi-lang3-4k.csv")

# Muestro las columnas y tipo de datos de cada una en cada dataframe
print("Columnas y tipos de datos del df_01:")
print(df_01.dtypes)
print("\n#############################")
print("Columnas y tipos de datos del df_02:")
print(df_02.dtypes)
print("\n#############################")
print("Columnas y tipos de datos del df_03:")
print(df_03.dtypes)

# Muestro los idiomas de cada dataframe
print("\nIdiomas del df_01:")
print(df_01['language'].unique())
print("\nIdiomas del df_02:")
print(df_02['language'].unique())
print("\nIdiomas del df_03:")
print(df_03['language'].unique())

# Elimino de todos los datasets los registros que no sean en español o inglés
df_01 = df_01[df_01['language'].isin(['es', 'en'])]
df_02 = df_02[df_02['language'].isin(['es', 'en'])]
df_03 = df_03[df_03['language'].isin(['es', 'en'])]
```

### Ejecución de Limpieza Estructural (df_01)

Finalizada la fase de auditoría exploratoria, se procede con la estandarización estructural del primer bloque de datos (`df_01`) para permitir su futura unificación.

**Modificaciones aplicadas:**
* **Selección de características:** Se conservan exclusivamente las columnas de interés analítico (`body`, `subject`, `answer`, `queue`, `type`, `priority`, `language`). Las columnas relacionadas con etiquetas y metadatos de versión se descartan por redundancia.
* **Filtrado de dominio:** Se excluyen las observaciones correspondientes a colas fuera del alcance de soporte técnico y facturación, con el fin de evitar sesgos en el entrenamiento de los algoritmos.
* **Tratamiento de nulos:** Se imputa un valor estático en campos vacíos de la variable de asunto.
* **Reindexación:** Se restablece el índice del dataframe para consolidar las eliminaciones de registros.

```python
# Eliminamos la columna 'version' del df_01
df_01 = df_01.drop(columns=['version'])

# Eliminamos todas las columnas tag del df_01
df_01 = df_01.drop(columns=[col for col in df_01.columns if col.startswith('tag_')])

# De la columna queue del df_01 eliminamos las filas donde aparezca Human Resources, Returns and Exchanges y General Inquiry
df_01 = df_01[~df_01['queue'].isin(['Human Resources', 'Returns and Exchanges', 'General Inquiry'])]

# En la columna 'subject' del df_01 rellenamos los valores nulos con el valor 'No Subject'
df_01['subject'] = df_01['subject'].fillna('No Subject')


# Reseteamos el indice del df_01 después de eliminar filas
df_01 = df_01.reset_index(drop=True)

# Mostramos el número total de filas y columnas actuales
print(f"\nNúmero total de filas y columnas actuales del df_01: {df_01.shape}")

# Mostramos el conteo de valores nulos
print("\nConteo de valores nulos despues de la limpieza del df_01:")
print(df_01.isnull().sum())

# Mostramos la distribución de frecuencias actualizada de la columna queue
print("\nDistribución de frecuencias de la columna 'queue' después de la limpieza:")
print(df_01['queue'].value_counts())

# Mostramos el primer y último índice del df_01 después de la limpieza
print(f"\nPrimer índice del df_01 después de la limpieza: {df_01.index}")


```

### Ejecución de Limpieza Estructural (df_02)

Se procede con la estandarización del segundo bloque de datos (`df_02`) para asegurar la coherencia esquemática con el dataset previo.

**Modificaciones aplicadas:**
* **Selección de características:** Se eliminan las variables de etiquetado por falta de aportación analítica.
* **Filtrado de dominio:** Se excluyen los registros que no pertenecen a las categorías objeto de estudio.
* **Tratamiento de nulos:** Imputación de nulos en la columna de asunto y eliminación directa de observaciones sin cuerpo de texto, dado que esta variable es un requisito crítico para la fase de vectorización semántica.
* **Reindexación:** Consolidación de los índices tras las alteraciones.

```python
# Elimina cualquier columna cuyo nombre empiece por la cadena de texto tag_.
df_02 = df_02.drop(columns=[col for col in df_02.columns if col.startswith('tag_')])

# Nos quedamos únicamente con las filas cuya columna queue no contenga las tres categorías descartadas (`Human Resources`, `Returns and Exchanges`, `General Inquiry`)
df_02 = df_02[~df_02['queue'].isin(['Human Resources', 'Returns and Exchanges', 'General Inquiry'])]

# Rellenamos los nulos de la columna subject con el valor 'No Subject'.
df_02['subject'] = df_02['subject'].fillna('No Subject')

# Eliminamos cualquier fila que tenga un nulo en la columna body.
df_02 = df_02.dropna(subset=['body'])

# Resetamos el índice del dataframe después de eliminar filas.
df_02 = df_02.reset_index(drop=True)

# Las dimensiones finales: Necesito ver la tupla con el número resultante de filas y el número exacto de columnas.
print(f"Dimensiones finales del df_02: {df_02.shape}")

# COmprobamos los nulos de cada columna del df_02
print("\nConteo de valores nulos del df_02:")   
print(df_02.isnull().sum())

# Comprobamos los valores unicos de la columna queue del df_02
print("\nValores únicos de la columna 'queue' del df_02 después de la limpieza:")
print(df_02['queue'].unique())
```

### Ejecución de Limpieza Estructural (df_03)

Se ejecuta el protocolo de saneamiento sobre el tercer bloque de datos (`df_03`) para forzar la compatibilidad dimensional estricta con los conjuntos anteriores.

**Modificaciones aplicadas:**
* **Selección de características:** Eliminación de variables de etiquetado y columnas intrusas ajenas a las siete características definidas para el proyecto.
* **Filtrado de dominio:** Segmentación de colas para focalizar la muestra en el área de estudio.
* **Tratamiento de nulos:** Imputación estática en el asunto y exclusión de observaciones carentes de cuerpo de mensaje.
* **Reindexación:** Ajuste estructural del índice para la concatenación posterior.

```python
# Elminamos las columnas que empiecen por tag_
df_03 = df_03.drop(columns=[col for col in df_03.columns if col.startswith('tag_')])

# Eliminamos la columna business_type
df_03 = df_03.drop(columns=['business_type'])

# Eliminamos filas donde la columna queue contenga Human Resources, Returns and Exchanges y General Inquiry
df_03 = df_03[~df_03['queue'].isin(['Human Resources', 'Returns and Exchanges', 'General Inquiry'])]

# Rellenamos los nulos de la columna subject con el valor 'No Subject'.
df_03['subject'] = df_03['subject'].fillna('No Subject')

# Eliminamos filas con nulos en body
df_03 = df_03.dropna(subset=['body'])

# Reseteamos el índice
df_03 = df_03.reset_index(drop=True)

# Mostramos las dimensiones finales del dataframe
print(f"\nDimensiones finales del df_03: {df_03.shape}")

# Hacemos recuentos de nulos de cada columna del df_03
print("\nConteo de valores nulos del df_03:")
print(df_03.isnull().sum())

# COmprobamos cuantas filas hay en idioma español
print("\nNúmero de filas en idioma español en el df_03:")
print(df_03[df_03['language'] == 'es'].shape[0])

```

### Consolidación de Datasets (Capa Silver)

```python
# Vamos a unir en df_final los tres datasets
df_final = pd.concat([df_01, df_02, df_03], ignore_index=True)

# Mostramos las dimensiones del df_final
print(f"\nNúmero total de filas y columnas del df_final: {df_final.shape}")

# Mostramos el conteo de valores nulos de cada columna del df_final
print(f"\nConteo de valores nulos del df_final:")
print(df_final.isnull().sum())

# Mostramos el número de filas duplicadas en el df_final
print(f"\nNúmero de filas duplicadas en el df_final: {df_final.duplicated().sum()}")

#Mostramos el numero de filas por idioma en el df_final
print(f"\nNúmero de filas por idioma en el df_final:")
print(df_final['language'].value_counts())

# Mostramos la distribución de frecuencias de la columna queue del df_final
print(f"\nDistribución de frecuencias de la columna queue del df_final:")
print(df_final['queue'].value_counts(normalize=True) * 100)
```

```python
# Eliminamos las filas duplicadas del df_final
df_final = df_final.drop_duplicates()

# Mostramos las dimensiones del df_final después de eliminar duplicados
print(f"\nNúmero total de filas y columnas del df_final después de eliminar duplicados: {df_final.shape}")

# Mostramos el conteo de valores nulos de cada columna del df_final
print(f"\nConteo de valores nulos del df_final:")
print(df_final.isnull().sum())

# Mostramos el número de filas duplicadas en el df_final
print(f"\nNúmero de filas duplicadas en el df_final: {df_final.duplicated().sum()}")

#Mostramos el numero de filas por idioma en el df_final
print(f"\nNúmero de filas por idioma en el df_final:")
print(df_final['language'].value_counts())

# Mostramos la distribución de frecuencias de la columna queue del df_final
print(f"\nDistribución de frecuencias de la columna queue del df_final:")
print(df_final['queue'].value_counts(normalize=True) * 100)

# Guardamos en data/processed el df_final en formato parquet
df_final.to_parquet ('../data/processed/df_final_silver.parquet', index=False)
```
